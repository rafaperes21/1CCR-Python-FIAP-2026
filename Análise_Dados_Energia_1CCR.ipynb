```python
# RENOMEAR OS RÓTULOS DOS ATRIBUTOS
# PASSAR COMO UM PARÂMETRO UM DICIONÁRIO QUE TEM:
# Chaves --> rótulos atuais
# Valores --> rótulos desejados
dados = dados.rename({'Date': 'Data',
                      'Time': 'Hora' ,
                      'Global_active_power': 'Potencia_ativa' ,
                      'Global_reactive_power': 'Potencia_reativa',
                      'Voltage': 'Tensão',
                      'Global_intensity': 'Corrente' ,
                      'Sub_metering_1': 'Consumo1',
                      'Sub_metering_2': 'Consumo2' ,
                      'Sub_metering_3': 'Consumo3' }, axis=1)
```

```python
dados.columns
```

```python
df1 = dados.drop(columns=['Data', 'Hora'])
df1.head()
```

```python
# SOMENTE UMA COLUNA ----> Pandas Serie
corrente = dados['Corrente']
corrente.head()
```

```python
type(corrente)
```

```python
# CRIAR UM DATAFRAME 'df2' A PARTIR DE 'df1' SOMENTE COM AS COLUNAS DE TENSAO,
# CORRENTE E POTENCIAS
df2 = df1[['Tensão', 'Corrente', 'Potencia_ativa', 'Potencia_reativa']]
df2.head()
```

```python
type(df2)
```

```python
# CRIAR UM DATAFRAME 'df31 A PARTIR DE 'dados' SOMENTE COM AS 4 PRIMEIRAS COLUNAS
# iloc
# : --------------> todos os registros (linhas)
# 0:4 ----------> da coluna de indice 0 até a coluna anterior à coluna 4
# ilocr ------> por indice
# loc -------> por rotulos
df3 = dados.iloc[:, -4:]
df3.head()
```

```python
# Qual é o valor máximo da demanda energética (Potencia_ativa) ----> PMAX
PMAX = dados["Potencia_ativa"].max()
print(f'O valor máximo da demanda energética é {PMAX:.2f} KW')
```

```python
# Qual o limiar de 60% da demanda energética máxima? ----> P60
P60 = PMAX * 0.6
print(f'O limiar de 60% de demanda energética máxima é {P60:.2f} KW')
```

```python
# Criar um dataframe 'df4' a partir de 'dados somente com as colunas
# Data, Hora, Corrente e Potencia ativa
df4 = dados[['Data', 'Hora', 'Corrente', 'Potencia_ativa']]
df4.head()
```

```python
# Criar um dataframe 'df5' a partir de 'df4' somente com os registros e consumidores
# com demanda acima de P60
df5 = df4[df4['Potencia_ativa'] > P60]
df5.head()
```
